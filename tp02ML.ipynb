```python
import pandas as pd
fruits =pd.Series([20, 15, 40, 30, 25], index=["Apple", "Banana", "Orange", "Mango", "Grapes"])
print(fruits)
third_fruit_price=fruits.iloc[2]
print(third_fruit_price)
new_price =fruits * 1.1
print(new_price)

average_price = new_price.mean()

most_expensive_fruit = new_price.idxmax()
highest_price = new_price.max()

print("Average price:", average_price)
print("Most expensive fruit:", most_expensive_fruit)
print("Highest price:", highest_price)
```

```python
import pandas as pd

data = {
    "Product": ["A", "B", "A", "C", "B", "A", "C", "B"],
    "Region": ["North", "South", "East", "West", "North", "South", "East", "West"],
    "Sales": [100, 200, 150, 300, 250, 120, 280, 180]
}

df = pd.DataFrame(data)

print(df)

total_sales = df.groupby("Product")["Sales"].sum()

print(total_sales)

average_sales = df.groupby("Region")["Sales"].mean()

print(average_sales)
sales_summary = df.groupby("Product")["Sales"].agg(["mean", "max"])

print(sales_summary)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Quarter": ["Q1", "Q2", "Q3", "Q4"],
    "Revenue": [12000, 15000, 11000, 18000]
}

df = pd.DataFrame(data)

print(df)

plt.plot(df["Quarter"], df["Revenue"], marker="o")

plt.title("Quarterly Revenue")
plt.xlabel("Quarter")
plt.ylabel("Revenue")

plt.show()

plt.bar(df["Quarter"], df["Revenue"])

plt.title("Revenue Comparison Across Quarters")
plt.xlabel("Quarter")
plt.ylabel("Revenue")

plt.show()

students = {
    "Student": ["Alice", "Bob", "Charlie", "David", "Eve",
                "Frank", "Grace", "Helen", "Jack", "Kate"],
    "Score": [55, 60, 65, 70, 72, 75, 80, 85, 90, 95]
}

df_students = pd.DataFrame(students)

print(df_students)

plt.hist(df_students["Score"], bins=5, edgecolor="red")

plt.title("Distribution of Student Test Scores")
plt.xlabel("Score")
plt.ylabel("Number of Students")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


df = sns.load_dataset("iris")
virginica = df[df["species"] == "virginica"]

print(virginica.head(5))
min_width = df["petal_width"].min()
max_width = df["petal_width"].max()

print("Minimum petal width:", min_width)
print("Maximum petal width:", max_width)
average_length = df.groupby("species")["sepal_length"].mean()

print(average_length)

highest_species = average_length.idxmax()

print("Species with highest average sepal length:", highest_species)
sns.scatterplot(
    data=df,
    x="petal_length",
    y="petal_width",
    hue="species"
)

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")

plt.show()
average_petal = df.groupby("species")["petal_length"].mean()

average_petal.plot(kind="bar")

plt.title("Average Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Average Petal Length (cm)")

plt.xticks(rotation=0)
plt.show()
```
